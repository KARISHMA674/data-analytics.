# ***4. corelation algorithm***

**basic correlation algo.**

```python
import numpy as np

x=np.array([1,2,3,4,5])
y=np.array([10,9,8,7,6])
def pearson_Correlation(x,y):
  if len(x)==len(y):
    sum_xy=sum((x-x.mean())*(y-y.mean()))
    sum_x_squared=sum((x-x.mean()**2))
    sum_y_squared=sum((y-y.mean()**2))
    corr=sum_xy/np.square(sum_x_squared*sum_y_squared)
  return corr
print(pearson_Correlation(x,y))
print(pearson_Correlation(x,x))
```

***5. Regression analysis algorithm***

```python
import matplotlib.pyplot as plt
from scipy import stats
x=[35,40,38,44,67,64,59,69,25,50]
y=[112,128,130,138,158,162,140,175,125,142]
slope,intercept,r,p,std_err=stats.linregress(x,y)
def myfunc(x):
  return slope*x+intercept
mymodel=list(map(myfunc,x))
plt.scatter(x,y)
plt.plot(x,mymodel)
plt.show()
print (slope)
print(intercept)
print(r)
print(p)
print(std_err)
```

**plot of nornmal distribution ,+ve skweness and -ve skweness.**

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import skew,skewnorm
sample = 1000
R_Skew=skewnorm.rvs(a=10,size=sample)
L_Skew=skewnorm.rvs(a=-10,size=sample)
normal=np.random.normal(0,1,sample)
fig,axes = plt.subplots(1,3,figsize=(15,5))
titles = ["positive Skew(Right)",'Normal Distribution','Negative Skew']
datasets = [R_Skew,normal,L_Skew]
for i,data in enumerate (datasets):
  sns.histplot(data,kde=True,ax=axes[i],color='skyblue',element='step')
  s_val= skew(data)
  mean_val=np.mean(data)
  median_val=np.median(data)
  axes[i].axvline(mean_val,color='red',linestyle ='--')
  axes[i].axvline(median_val,color='green',linestyle='--')
plt.show()
```

**T-TEST**
Compare the mean of the two groups or to compare a sample mean to a known population mean when the population standard deviation is unknown and the sample size is relatively small.

```python
from scipy import stats
import numpy as np
sample_data=np.array([10,12,11,13,10,9,12,11,10,13])
population_mean=10.5
t_statistics,p_value=stats.ttest_1samp(sample_data,population_mean)
print(f"one-sample t-test: t-statistics={t_statistics:.2f}, p_value={p_value:.2f})")
```

if two or more group_sample use independent t test.

```python
group1=np.array([25,28,27,26,29])
group2=np.array([22,24,23,21,25])
t_statistics,p_value=stats.ttest_ind(group1,group2)
print(f"Two sample t-test : t_statistics={t_statistics : .2f}, p_value:{p_value: .3f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
np.random.seed(42)
group_a=np.random.normal(loc=50,scale=10,size=100)
print(group_a)
group_b=np.random.normal(loc=55,scale=10,size =100)
print(group_b)
S_statistics,p_value=stats.ttest_ind(group_a,group_b,equal_var=True)
plt.figure(figsize=(10,6))
sns.histplot(group_a,color='green',label='GroupA',kde=True,stat='density',alpha=0.6)
sns.histplot(group_b,color='skyblue',label='GroupB',kde=True,stat='density',alpha=0.6)
print(f"t-statistics={S_statistics:.2f}, p_value={p_value:.3f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
np.random.seed(42)
group_a=np.random.normal(loc=50,scale=10,size=100)
print(group_a)
group_b=np.random.normal(loc=55,scale=10,size =100)
print(group_b)
S_statistics,p_value=stats.ttest_ind(group_a,group_b,equal_var=True)
plt.figure(figsize=(10,6))
sns.histplot(group_a,color='green',label='GroupA',kde=True,stat='density',alpha=0.6)
sns.histplot(group_b,color='skyblue',label='GroupB',kde=True,stat='density',alpha=0.6)
print(f"t-statistics={S_statistics:.2f}, p_value={p_value:.3f}")
plt.axvline(np.mean(group_a),color='blue',linestyle='--',linewidth=2,label='Mean groupA')
plt.axvline(np.mean(group_b),color='red',linestyle='--',linewidth=2,label='Mean groupB')
plt.legend()
plt.show()
```

```python
import pandas as pd
import yfinance as yf
df1=yf.download("TATAPOWER.NS")
print(df1)
df1=pd.DataFrame(df1)
df1=df1[['Close']]
df1_p=df1.pct_change(1)
print(df1_p)

df2=yf.download("SNSE")
print(df2)
df2=pd.DataFrame(df2)
df2=df2[['Close']]
df2_p=df2.pct_change(1)
print(df2_p)
df3=pd.concat([df1_p,df2_p],axis=1)
print(df3)
print(df3.mean())
print(df3.median())
print(df3.corr())
print(df3.skew())
print(df3.kurt())
```
