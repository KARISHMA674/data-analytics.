# ***4. corelation algorithm***

**basic correlation algo.**

```python
import numpy as np

x=np.array([1,2,3,4,5])
y=np.array([10,9,8,7,6])
def pearson_Correlation(x,y):
  if len(x)==len(y):
    sum_xy=sum((x-x.mean())*(y-y.mean()))
    sum_x_squared=sum((x-x.mean()**2))
    sum_y_squared=sum((y-y.mean()**2))
    corr=sum_xy/np.square(sum_x_squared*sum_y_squared)
  return corr
print(pearson_Correlation(x,y))
print(pearson_Correlation(x,x))
```

**different types of test:-(5)**
1. Z-Test: Large sample, known population standard deviation, compare means.
example-> Battery life of 1,000 phones.
2. T-Test: Small sample, unknown population standard deviation, compare means.
example->Marks of 20 students.
3. F-Test: Compare variances or perform ANOVA.
example->Two machines' consistency.
4. P-Test: Test proportions or percentages.
example->Voter support percentage.
5. Chi-Square Test: Analyze relationships between categorical variables.
example->Gender vs Product Choice.

**T-TEST**
T-Test: Small sample, unknown population standard deviation, compare means. example->Marks of 20 students.

Compare the mean of the two groups or to compare a sample mean to a known population mean when the population standard deviation is unknown and the sample size is relatively small.

```python
from scipy import stats
import numpy as np
sample_data=np.array([10,12,11,13,10,9,12,11,10,13])
population_mean=10.5
t_statistics,p_value=stats.ttest_1samp(sample_data,population_mean)
print(f"one-sample t-test: t-statistics={t_statistics:.2f}, p_value={p_value:.2f})")
```

**Z-test.**
Large sample, known population standard deviation, compare means.
example-> Battery life of 1,000 phones.

```python
from scipy import stats
import numpy as np
sample_mean=65
population_mean=60
population_std=5
sample_size=100
z_statisctics=(sample_mean-population_mean)/(population_std/np.sqrt(sample_size))
p_value=2*(1-stats.norm.cdf(np.abs(z_statisctics)))  #for a two tailed test #continuous density function #absolute value
print(f"z_test:z_score={z_statisctics:.2f},p_value={p_value:.3f}")
```

```python
from scipy.stats import norm
import numpy as np
import matplotlib.pyplot as plt # Import matplotlib.pyplot
z_statisctics=1.96  #z_statistics
alpha=0.5  #significance level for two tail test
x=np.linspace(-4,4,500)
pdf=norm.pdf(x,0,1)  #mean=0,S.D=1 #probability density function-pdf
plt.figure(figsize=(10,6))
plt.plot(x,pdf,color='blue',label='STND')
plt.title('z-test visualization')
plt.xlabel('z_score')
plt.ylabel('probability Density')
plt.grid(True)
plt.show() # Corrected typo from plt.shoow() to plt.show()
```

```python
#add critical region (two tailed test)
critical_value_upper=norm.ppf(1-(alpha/2))
critical_value_lower=norm.ppf(alpha/2)
plt.axvline(critical_value_upper)
plt.axvline(critical_value_lower)
plt.show()
```

```python
#add z statistics
plt.axvline(z_statisctics,color='red',linestyle='--',label=f'calculated z_statistic:{z_statisctics:.2f}')
```

```python
#shade the rejection
x_reject_upper=x[x>critical_value_upper]
plt.fill_between(x_reject_upper,0,norm.pdf(x_reject_upper,0,1))
x_reject_lower=x[x>critical_value_lower]
plt.fill_between(x_reject_lower,0,norm.pdf(x_reject_lower,0,1))
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import f
dfn=5 #numerator degree of freedom
dfd=10 #denomenator degree of freedom
alpha = 0.05 #signifinance level
critical_value=f.ppf(1-alpha,dfn,dfd)
X=np.linspace(0,f.ppf(0.999,dfn,dfd),500)
print(X)
#calculate pdf
pdf=f.pdf(X,dfn,dfd)
calculate_f_statistic=2.5
plt.figure(figsize=(10,6))
plt.plot(X,pdf,label=f'F-distribution(dfn={dfn},dfd={dfd}')

#reject
X_reject=np.linspace(critical_value,max(X),100)
plt.fill_between(X_reject,0,f.pdf(X_reject,dfn,dfd),color='red',alpha=0.3)
plt.axvline(critical_value,color='red',linestyle='--',label=f'critical_value(a={alpha:.2f})')

plt.axvline(calculate_f_statistic,color='blue')
plt.show()
```

## F-test for Equality of Variances (Levene's Test)

The F-test can be used to compare the variances of two populations. A common application in finance is to compare the volatility (represented by variance) of two different assets.

Here, we will use **Levene's test**, which is a robust alternative to the classic F-test for equality of variances, as it is less sensitive to departures from normality.

**Hypotheses:**
*   **Null Hypothesis (H0):** The variances of the two populations are equal.
*   **Alternative Hypothesis (H1):** The variances of the two populations are not equal.

```python
import pandas as pd
import yfinance as yf
from scipy.stats import levene
import numpy as np

# Re-download data for BTC-USD and MSFT to ensure we have the correct dataframes
df_btc = yf.download("BTC-USD", period="1mo")
df_msft = yf.download("MSFT", period="1mo")

# Calculate daily percentage change for 'Close' prices
btc_returns = df_btc['Close'].pct_change().dropna()
msft_returns = df_msft['Close'].pct_change().dropna()

# Ensure both series have data to compare
if len(btc_returns) > 1 and len(msft_returns) > 1:
    # Perform Levene's test
    statistic, p_value = levene(btc_returns, msft_returns)

    print(f"Levene's Test for Equality of Variances:")
    print(f"Statistic = {statistic.item():.4f}")
    print(f"P-value = {p_value.item():.4f}")

    alpha = 0.05
    if p_value.item() < alpha:
        print(f"Conclusion: Reject the null hypothesis. There is a significant difference in variances (volatility) between BTC-USD and MSFT (p < {alpha}).")
    else:
        print(f"Conclusion: Fail to reject the null hypothesis. There is no significant difference in variances (volatility) between BTC-USD and MSFT (p >= {alpha}).")
else:
    print("Not enough data to perform Levene's test for one or both assets.")
```

To visualize the distributions and compare their variances, we can plot histograms and box plots of the daily returns.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import yfinance as yf

# Re-download data for BTC-USD and MSFT to ensure we have the correct dataframes
df_btc = yf.download("BTC-USD", period="1mo")
df_msft = yf.download("MSFT", period="1mo")

# Calculate daily percentage change for 'Close' prices
btc_returns = df_btc['Close'].pct_change().dropna()
msft_returns = df_msft['Close'].pct_change().dropna()

plt.figure(figsize=(14, 6))

# Histogram of returns
plt.subplot(1, 2, 1)
sns.histplot(btc_returns, color='blue', label='BTC-USD Returns', kde=True)
sns.histplot(msft_returns, color='green', label='MSFT Returns', kde=True)
plt.title('Distribution of Daily Percentage Changes')
plt.xlabel('Daily Percentage Change')
plt.ylabel('Frequency')
plt.legend()

# Box plot of returns
plt.subplot(1, 2, 2)
sns.boxplot(data=[btc_returns.squeeze(), msft_returns.squeeze()], palette=['blue', 'green'])
plt.xticks([0, 1], ['BTC-USD', 'MSFT'])
plt.title('Box Plot of Daily Percentage Changes')
plt.ylabel('Daily Percentage Change')

plt.tight_layout()
plt.show()
```

**bargraph plot of above data of chi-square.**

```python
import pandas as pd
import numpy as np
import sklearn
from sklearn.feature_selection import SelectKBest,chi2
import matplotlib.pyplot as plt
data={'feature1' :[10,20,15,25,30],
'feature2' :[5,12,8,18,22],
'feature3' :[3,7,4,9,11],
'target' :[0,1,0,1,0]}
df=pd.DataFrame(data)
x=df[['feature1','feature2','feature3']]
y=df['target']
#chi-square test
selector=SelectKBest(chi2,k='all')
selector.fit(x,y)
chi_scores=pd.Series(selector.scores_,index=x.columns)
chi_scores=chi_scores.sort_values(ascending=False)
#plot
plt.figure(figsize=(8,6))
chi_scores.plot(kind='bar')
plt.title('feature Importance basedon chi-square scores')
plt.xlabel('features')
plt.ylabel('chi-square scores')
plt.xticks(rotation=45,ha='right')
plt.tight_layout()
plt.show()
print(x)
print(y)
print(selector)
print(chi_scores)
```
